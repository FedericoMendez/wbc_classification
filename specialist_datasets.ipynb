```python
import pandas as pd
import os


# -------------------------
# DEFINE CLUSTERS
# -------------------------
CLUSTERS = {
    "myeloid": ["MY", "MMY", "PMY"],
    "lymphoid": ["LY", "VLY", "PLY"],
    "neutrophil": ["BNE", "SNE"]
}


# -------------------------
# FUNCTION
# -------------------------
def create_cluster_csvs(train_csv, val_csv, output_dir="cluster_splits"):
    """
    Creates cluster-specific train/val CSVs for each cluster.

    Args:
        train_csv (str): path to full train CSV
        val_csv (str): path to full val CSV
        output_dir (str): where to save cluster CSVs
    """

    os.makedirs(output_dir, exist_ok=True)

    # Load full datasets
    train_df = pd.read_csv(train_csv)
    val_df   = pd.read_csv(val_csv)

    for cluster_name, cluster_classes in CLUSTERS.items():

        # Filter train
        train_cluster = train_df[train_df["label"].isin(cluster_classes)]

        # Filter val
        val_cluster = val_df[val_df["label"].isin(cluster_classes)]

        # Save
        train_path = os.path.join(output_dir, f"train_{cluster_name}.csv")
        val_path   = os.path.join(output_dir, f"val_{cluster_name}.csv")

        train_cluster.to_csv(train_path, index=False)
        val_cluster.to_csv(val_path, index=False)

        print(f"✅ {cluster_name}")
        print(f"   Train: {len(train_cluster)} samples → {train_path}")
        print(f"   Val:   {len(val_cluster)} samples → {val_path}")
        print()


# -------------------------
# RUN
# -------------------------
if __name__ == "__main__":
    create_cluster_csvs(
        train_csv="train_split.csv",
        val_csv="val_split.csv",
        output_dir="cluster_splits"
    )
```
